```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_parquet('../data/processed/features.parquet')
latest = df[df.cutoff == df.cutoff.max()].copy()
cols = ['matches_30d', 'avg_duration_30d', 'kda_30d', 'win_rate_30d']
X = StandardScaler().fit_transform(latest[cols].fillna(latest[cols].median()))
latest['segment'] = KMeans(n_clusters=4, n_init=10, random_state=0).fit_predict(X)
print(latest.groupby('segment')[cols].mean())   # name each segment from these averages
```
